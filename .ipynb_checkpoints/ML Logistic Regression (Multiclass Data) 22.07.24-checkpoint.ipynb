**Dataset Overview**

The **Iris dataset** is a classic dataset frequently used in the fields of machine learning and statistics. It was first introduced by Ronald Fisher in 1936 and is commonly used to distinguish between three different species of Iris flowers (Iris setosa, Iris virginica, and Iris versicolor). This dataset is widely used for training and testing classification algorithms in machine learning and statistical applications.

**Details of the Iris Dataset:**

The Iris dataset contains a total of 150 flower samples, with each species of Iris flower represented by 50 samples:

- **Iris-setosa**
- **Iris-versicolor**
- **Iris-virginica**

For each flower sample, four features have been recorded:

1. **sepal_length**: The length of the sepal (the outer part of the flower that protects the petals).
2. **sepal_width**: The width of the sepal.
3. **petal_length**: The length of the petal (the colorful, showy part of the flower).
4. **petal_width**: The width of the petal.

These features are measured in centimeters (cm).

**Objective:**

The objective is to create a model that can predict the species of the flower using these four features. In other words, the Iris dataset is used as a dataset for a supervised learning problem. The goal of the model is to accurately classify the flower species based on the given features.

**Summary:**

The Iris dataset contains 150 flower samples, each representing one of three Iris species. Four features (sepal_length, sepal_width, petal_length, petal_width) have been measured for each sample. These features are used to distinguish between the different Iris species.

- **Sepal Length:** The length of the sepal.
- **Sepal Width:** The width of the sepal.
- **Petal Length:** The length of the petal.
- **Petal Width:** The width of the petal.

This dataset is an ideal example for evaluating and comparing the performance of classification algorithms.

# Train | Test Split

**Stratify Parametresi Hakkında Bilgi**

**Stratify** parametresi, Scikit-Learn kütüphanesindeki **train_test_split** gibi fonksiyonlarda kullanılan ve veri setini eğitim ve test setlerine bölerken hedef değişkenin sınıf dağılımını korumak için kullanılan bir parametredir. Bu, özellikle sınıflandırma problemlerinde, sınıflar arasında dengesizlik olduğunda önemlidir.

**Stratify Kullanmanın Önemi**

Dengesiz veri setleriyle çalışırken, veriyi rastgele olarak ayırmak, eğitim ve test setlerinde sınıfların orijinal dağılımını yansıtmayabilir. Bu durum, modelin performansının ve değerlendirme metriklerinin güvenilmez olmasına yol açabilir. Stratify parametresi kullanılarak, her iki sette de sınıfların orantılı olarak temsil edilmesi sağlanır.

**Stratify Parametresinin Faydaları**

1. **Sınıf Dağılımını Korur:** Eğitim ve test setlerindeki sınıfların orijinal veri setindeki oranlarını korur.
2. **Model Değerlendirmesini İyileştirir:** Modelin performansının, görülmemiş veriler üzerinde daha doğru bir şekilde değerlendirilmesini sağlar.
3. **Önyargıyı Önler:** Modelin, çoğunluk sınıfta iyi performans gösterip, azınlık sınıfta kötü performans göstermesini engeller.

**Stratify Parametresinin Kullanımı**

Stratify parametresi, hedef değişken (genellikle **y**) olarak belirlenir. Bu sayede, verilerin eğitim ve test setlerine ayrılırken hedef değişkenin sınıf dağılımı korunmuş olur. Örneğin, eğer orijinal veri setinde sınıflar %70-%30 oranında ise, stratify kullanarak yapılan ayırmada da aynı oran korunur.

**Sonuç**

Stratify parametresinin kullanımı, veri setlerini bölerken sınıfların dağılımını koruyarak, özellikle dengesiz sınıflar içeren veri setlerinde daha doğru ve güvenilir modeller elde edilmesine yardımcı olur. Bu, sınıflandırma problemlerinde modelin performansını değerlendirmek için en iyi uygulamalardan biridir.

# Modelling and Model Performance

**Çok Sınıflı (Multiclass) Problemler için Metrikler**

1. **Macro Average (Macro Avg):** 
    - Her sınıf için F1 puanı (ya da başka bir metrik) hesaplanır.
    - Bu değerler toplanır ve sınıf sayısına bölünerek ortalama alınır.
    - Macro average, tüm sınıfların eşit derecede önemli olduğu durumları temsil eder.
    - Dengesiz veri setlerinde, az örneğe sahip sınıfların performansı da aynı derecede dikkate alınır.
    - Özellikle tüm sınıfların eşit olarak değerlendirilmesini istediğinizde bu metriği kullanabilirsiniz.

2. **Weighted Average (Weighted Avg):** 
    - Her sınıf için F1 puanı (ya da başka bir metrik) hesaplanır.
    - Bu değerler, her sınıftaki örnek sayısıyla ağırlıklandırılarak ortalaması alınır.
    - Weighted average, sınıfların örnek sayısına bağlı olarak önemini dikkate alır.
    - Çokça örneğe sahip sınıfların performansı daha ağırlıklı olarak değerlendirilir.
    
3. **Micro Average (Micro Avg):**     

    - Tüm sınıflardaki doğru pozitif (TP), yanlış pozitif (FP) ve yanlış negatif (FN) sayıları toplanarak global bir F1 puanı hesaplanır.
    - Precision, recall ve F1 skoru gibi metrikler, sınıf bazında değil, global olarak hesaplanır.
    - Tüm sınıflardaki tahminlerin toplamı üzerinden hesap yapılır.
    - Çoğunluk sınıfının performansına daha fazla ağırlık verir.
    - Genel model performansını ölçmek için kullanışlıdır, özellikle veri seti dengeli olduğunda.

- Eğer tüm sınıfların performansının eşit derecede önemli olduğunu düşünüyorsanız, "Macro Avg" metriğine bakmalısınız.
- Ancak, bazı sınıfların daha fazla örneğe sahip olduğu ve bu sınıfların performansının daha kritik olduğunu düşünüyorsanız "Weighted Avg" metriğine odaklanmalısınız.

- **Mikro Recall**: Bu yöntem, tüm sınıflar üzerinden yapılan tahminlerin toplamını alarak tek bir toplu metrik hesaplar. Yani, tüm sınıfların doğru pozitif (TP) ve yanlış negatif (FN) değerlerinin toplamını kullanır. Bu, modelin tüm sınıflar arasında genel olarak nasıl performans gösterdiğini gösterir. Daha çok, sınıf dengesizliklerinin önemli olmadığı durumlarda tercih edilir.

- **Makro Recall**: Her bir sınıf için ayrı ayrı recall hesaplar ve sonra bu recall değerlerinin basit ortalamasını alır. Bu yaklaşım, tüm sınıfları eşit önemde tutar ve her bir sınıfın performansını eşit ağırlıkta değerlendirir. Sınıf dengesizliğinin olduğu durumlarda bu yöntem tercih edilebilir çünkü azınlık sınıflarının performansını göz ardı etmez.

**Genel olarak Mikro ve Makro Arasındaki Fark**

- **Mikro düzeydeki metrikler**, modelin tüm sınıflar arasındaki toplu performansına odaklanır. Bu, modelin genel başarımını gösterir ancak sınıflar arası dengesizlikleri göz ardı edebilir.
  
- **Makro düzeydeki metrikler**, her bir sınıfın performansını ayrı ayrı değerlendirir ve bu değerlerin ortalamasını alır. Bu yaklaşım, tüm sınıfların performansını eşit önemde değerlendirir, bu nedenle azınlık sınıfların performansı da büyük sınıflar kadar önem kazanır.

**Hangisini Ne Zaman Kullanmalıyım?**

- **Mikro düzey metrikleri** kullanmanız gereken durumlar, genel model performansını, özellikle de sınıf dengesizliğinin önemli olmadığı durumlarda ölçmek istediğinizde olabilir.

- **Makro düzey metrikleri** kullanmanız gereken durumlar ise, tüm sınıfların performansını eşit önemde değerlendirmek istediğinizde, özellikle de azınlık sınıfların performansını önemseyerek dengesiz veri setlerinde çalışırken olabilir.

**Ne Zaman Macro Skorlara Bakıyoruz, Ne Zaman Accuracy'ye Bakıyoruz?**

Dengesiz veri setleri, bazı sınıfların diğerlerine göre çok daha fazla örneğe sahip olduğu durumlardır. Örneğin, bir hastalığı teşhis eden bir modelde, hastalığa sahip olmayan (negatif) örneklerin sayısı, hastalığa sahip (pozitif) örneklerden çok daha fazla olabilir. Bu durumda, modelin çoğunluk sınıfını (negatif örnekler) öğrenmesi ve azınlık sınıfını (pozitif örnekler) göz ardı etmesi riski vardır. Bu tür dengesiz veri setlerinde hangi metriklerin kullanılması gerektiğini anlamak önemlidir.

**Macro Metrikler**

Dengesiz veri setlerinde, "macro" metriklere bakmak gerekir. Macro metrikler, her sınıf için ayrı ayrı hesaplanan performans metriklerinin basit bir şekilde ortalamasını alır. Örneğin, "macro F1 skoru", her sınıf için ayrı ayrı hesaplanan F1 skorlarının aritmetik ortalamasıdır. Bu durum, azınlık sınıfının performansını göz ardı etmeyi zorlaştırır ve modelin tüm sınıfları dengeli bir şekilde öğrenmesini sağlar.

**Macro Metriklerin Özellikleri:**
- Her sınıfın performansını ayrı ayrı değerlendirir.
- Modelin tüm sınıfları dengeli bir şekilde öğrenip öğrenmediğini gösterir.
- Azınlık sınıfların performansını göz ardı etmeyi zorlaştırır.

**Accuracy (Doğruluk) Metrikleri**

Accuracy (doğruluk) metriği ise genellikle dengesiz veri setlerinde yanıltıcı olabilir. Eğer bir model, çoğunluk sınıfını sürekli doğru tahmin ederse, yüksek bir doğruluk skoru elde edebilir. Bu durum, modelin azınlık sınıfı üzerinde iyi performans gösterdiği anlamına gelmez. Dolayısıyla, dengesiz veri setlerinde doğruluk yerine "macro" metrikler tercih edilir.

**Accuracy Metriklerinin Sınırlamaları:**
- Dengesiz veri setlerinde yanıltıcı olabilir.
- Çoğunluk sınıfı doğru tahmin edilse bile azınlık sınıfının performansı göz ardı edilebilir.
- Modelin azınlık sınıfları öğrenmedeki başarısını yansıtmaz.

**Sonuç**

Dengesiz veri setlerinde model performansını değerlendirirken macro metrikler kullanmak, her sınıfın performansını ayrı ayrı değerlendirdiği ve azınlık sınıfların performansını göz ardı etmediği için daha doğru bir yaklaşım sağlar. Accuracy metriği ise dengesiz veri setlerinde yanıltıcı olabileceği için dikkatli kullanılmalıdır. 

Modelin tüm sınıfları üzerinde dengeli bir performans göstermesi isteniyorsa, macro metriklere odaklanmak en iyi yaklaşım olacaktır. Bu, modelin yalnızca çoğunluk sınıfında değil, aynı zamanda azınlık sınıflarında da iyi performans göstermesini sağlar ve böylece genel model performansına daha bütüncül bir bakış açısı sunar.

## With Default Parameters

**Iris Veri Seti ve Confusion Matrix Analizi**

Iris veri seti, çok sınıflı (multi-class) bir sınıflandırma problemi için kullanılan klasik bir veri setidir. Bu veri setinde, üç farklı Iris çiçeği türü (Iris setosa, Iris versicolor, Iris virginica) sınıflandırılmaktadır. Sınıflandırma performansının değerlendirilmesinde kullanılan confusion matrix, modelin her bir sınıfı ne kadar iyi ayırt ettiğini gösteren önemli bir araçtır.

**Confusion Matrix'in Yapısı**

Confusion matrix, 3x3'lük bir matristir ve her bir sınıfın diğer sınıflarla nasıl karıştırıldığını gösterir. Her satır gerçek sınıfı (actual class), her sütun ise tahmin edilen sınıfı (predicted class) temsil eder.

**Iris versicolor için Confusion Matrix Analizi**

- **True Positive (TP) için Iris versicolor**: Modelin gerçekten Iris versicolor olan örnekleri doğru bir şekilde Iris versicolor olarak sınıflandırması. Confusion matrix'te, Iris versicolor satırı ve sütunu kesiştiğindeki hücre, TP sayısını gösterir.
- **False Positive (FP) için Iris versicolor**: Modelin Iris versicolor olmayan örnekleri (yani Iris setosa veya Iris virginica) yanlışlıkla Iris versicolor olarak sınıflandırması. Confusion matrix'te, Iris versicolor sütunu ancak Iris versicolor dışındaki satırlar FP sayılarını gösterir.
- **False Negative (FN) için Iris versicolor**: Modelin gerçekten Iris versicolor olan örnekleri yanlışlıkla başka bir sınıfa ataması. Confusion matrix'te, Iris versicolor satırı ancak Iris versicolor dışındaki sütunlar FN sayılarını gösterir.

```python
operations = [("scaler", StandardScaler()), ("logistic", LogisticRegression())]

pipe_model = Pipeline(steps=operations)

pipe_model.fit(X_train, y_train)
```

```python
y_pred = pipe_model.predict(X_test)
y_pred
```

```python
ConfusionMatrixDisplay.from_estimator(pipe_model, X_test, y_test);
```

```python
def eval_metric(model, X_train, y_train, X_test, y_test):
    y_train_pred = model.predict(X_train)
    y_pred = model.predict(X_test)
    
    print("Test_Set")
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    print()
    print("Train_Set")
    print(confusion_matrix(y_train, y_train_pred))
    print(classification_report(y_train, y_train_pred))
```

```python
eval_metric(pipe_model, X_train, y_train, X_test, y_test)

# multiclass classification problemlerinde eğer bütün classlar aynı öneme sahipse
# macro, weighted veya accuracy scorelarına göre sonuçları değerlendirmek daha doğru olacaktır.

# macro avg: bütün classlara eşit şekilde davranır.
# Gözlem sayısı daha fazla olan classın scorelarının ön plana çıkmasını istiyorsak da weighted avg kullanılır.

# micro_score accuracy'e eşittir. Data dengeli ise kullanılabilir.
```

## Cross Validate

```python
from sklearn.model_selection import cross_validate

operations = [("scaler", StandardScaler()), ("logistic", LogisticRegression())]

model = Pipeline(steps=operations)

scores = cross_validate(model, X_train, y_train, scoring = ['accuracy', 'precision_weighted','recall_weighted',
                                                                   'f1_weighted'], cv = 10, return_train_score=True)
df_scores = pd.DataFrame(scores, index = range(1, 11))
df_scores.mean()[2:]

# multiclass datalarda CV veya gridsearchCV için precision, recall ve f1 metriclerini tek başına kullanmayız. Yukarıda olduğu
# gibi precision_weighted, precision_macro, precision_micro vb. yazmalıyız. Yoksa hata alırsınız. Spesifik olarak tek bir 
# sınıfa ait CV scorlarını görmek istiyorsak aşağıda gösterildiği gibi make_scorer fonksiyonunu kullanmalıyız.
```

## Cross Validate for Versicolor

```python
from sklearn.metrics import make_scorer
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score, roc_auc_score

f1_versicolor = make_scorer(f1_score, average=None, labels=["versicolor"])

precision_versicolor = make_scorer(precision_score,
                                   average=None,
                                   labels=["versicolor"])

recall_versicolor = make_scorer(recall_score,
                                average=None,
                                labels=["versicolor"])

scoring = {
    "f1_versicolor": f1_versicolor,
    "precision_versicolor": precision_versicolor,
    "recall_versicolor": recall_versicolor
}

# Multiclass datalarda hedef labelınız ne ise onun üzerinden CV scorlarını alabilirsiniz.
# Yine make_scorer fonksiyonunu kullanamk zorundayız. Data multi class olduğunda make_scorer fonksiyonu içine
# average, ve labels parametreleri mutlaka belirtilmelidir.
# Label isimleri liste şeklinde verilmelidir.
# Tek bir class için scorer tanımlıyorsak average: None yapılmalıdır.
```

```python
operations = [("scaler", StandardScaler()), ("logistic", LogisticRegression())]
model = Pipeline(steps=operations)

scores = cross_validate(model,
                        X_train,
                        y_train,
                        scoring=scoring,
                        cv=10,
                        return_train_score=True)
df_scores = pd.DataFrame(scores, index=range(1, 11))
df_scores.mean()[2:]
```

```python
y_pred = pipe_model.predict(X_test)
y_pred_proba = pipe_model.predict_proba(X_test)
y_pred_proba = np.round(y_pred_proba, 2)
```

```python
test_data = pd.concat([X_test, y_test], axis=1)
test_data["pred"] = y_pred
test_data["pred_proba_setosa"] = y_pred_proba[:,0]
test_data["pred_proba_versicolar"] = y_pred_proba[:,1]
test_data["pred_proba_virginica"] = y_pred_proba[:,2]
test_data.sample(10)

# multiclass datalarda 0.5 thresholdu aranmaz. Hangi classa ait olasılık yüksek ise o sınıfa atanır.
```

## With Best Parameters (GridsearchCV)

**Logistic Regresyon Hiperparametreleri**

1. **Penalty (penalty):**
   - **Tanım:** Regularization türünü belirler.
   - **Seçenekler:** 'l1', 'l2', 'elasticnet', 'none'
     - 'l1': Lasso regularization
     - 'l2': Ridge regularization (varsayılan)
     - 'elasticnet': L1 ve L2'nin birleşimi
     - 'none': Regularization uygulanmaz
   - **Kullanım:** Overfitting'i önlemek için kullanılır. L1 sparse modeller oluşturur ve bazı katsayıları sıfıra indirir. L2 ise tüm katsayıları küçültür ancak sıfır yapmaz.

2. **Inverse of Regularization Strength (C):**
   - **Tanım:** Regularization teriminin tersine karşılık gelir.
   - **Kullanım:** C değeri ne kadar küçükse, regularization o kadar güçlü olur. C değeri ne kadar büyükse, regularization o kadar zayıf olur.

3. **Solver (solver):**
   - **Tanım:** Optimizasyon problemini çözmek için kullanılan algoritma.
   - **Seçenekler:** 'newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'
     - 'newton-cg', 'lbfgs', 'sag', 'saga': L2 penalty ile çalışır.
     - 'liblinear': Küçük veri setleri için uygundur, L1 ve L2 penalty ile çalışır.
     - 'saga': Büyük veri setleri için uygundur, L1, L2, ve elasticnet penalty ile çalışır.
   - **Kullanım:** Veri setinin boyutu ve regularization türüne göre solver seçilir.

4. **Maximum Number of Iterations (max_iter):**
   - **Tanım:** Optimizasyon algoritmasının maksimum iterasyon sayısı.
   - **Kullanım:** Optimizasyon algoritmasının yakınsama sağlamak için izin verilen maksimum iterasyon sayısını belirler. Bazı durumlarda, yüksek değerler daha iyi performans sağlar.

5. **Class Weight (class_weight):**
   - **Tanım:** Sınıflar için ağırlıklandırma.
   - **Seçenekler:** 'balanced', {dict, 'balanced'}, None
     - 'balanced': Sınıf ağırlıkları otomatik olarak ayarlanır, sınıf dengesizliklerini düzeltir.
     - {dict}: Her sınıf için manuel olarak belirlenmiş ağırlıklar.
     - None: Ağırlıklandırma yapılmaz (varsayılan).
   - **Kullanım:** Sınıf dengesizliklerini ele almak için kullanılır.

7. **Multi-class (multi_class):**
   - **Tanım:** Çok sınıflı problemler için kullanılacak strateji.
   - **Seçenekler:** 'auto', 'ovr', 'multinomial'
     - 'auto': Veri setinin boyutuna göre otomatik seçim yapar.
     - 'ovr': One-vs-Rest (her sınıfa karşı diğerleri).
     - 'multinomial': Doğrudan çok sınıflı hedef.
   - **Kullanım:** Çok sınıflı sınıflandırma problemlerinde kullanılır.

**Özet**

- **Penalty (penalty):** Regularization türü ('l1', 'l2', 'elasticnet', 'none').
- **Inverse of Regularization Strength (C):** Regularization gücü (küçük değerler güçlü regularization).
- **Solver (solver):** Optimizasyon algoritması ('newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga').
- **Maximum Number of Iterations (max_iter):** Optimizasyonun maksimum iterasyon sayısı.
- **Class Weight (class_weight):** Sınıf ağırlıkları ('balanced', {dict}, None).
- **Multi-class (multi_class):** Çok sınıflı problemler için strateji ('auto', 'ovr', 'multinomial').

```python
from sklearn.model_selection import GridSearchCV
```

```python
operations = [("scaler", StandardScaler()), ("logistic", LogisticRegression())]

model = Pipeline(steps=operations)
```

```python
# l1: Lasso, l2: Ridge
penalty = ["l1", "l2"]

# -1 ile 5 arasında 20 tane C değeri elde etmek için
# C arttıkça regülarizasyon düşer, alphanın tersi olarak
# C için np.logspace kullanmanın nedeni, regülarizasyonun optimal değerinin genellikle belirli bir aralıkta olmamasıdır.
# Bu nedenle, geniş bir aralıkta farklı değerleri denemek ve en iyi sonucu veren değeri bulmak için logaritmik bir ölçekte
# değerler oluşturmak mantıklıdır.
C = np.logspace(-1, 5, 20)

# solver ın dört değeri
# solver = ["lbfgs", "liblinear", "sag", "saga"]

# penalty, C, class_weight ve solver'ın tüm kombinasyonlarını elde etmek için
param_grid = [
    {
        "logistic__penalty": ["l1"],
        "logistic__C": C,
        "logistic__solver": ["liblinear", "saga"]
    },
    {
        "logistic__penalty": ["l2"],
        "logistic__C": C,
        "logistic__solver": ["newton-cg", "lbfgs", "liblinear", "sag", "saga"]
    },
    {
        "logistic__penalty": ["elasticnet"],
        "logistic__C": C,
        "logistic__solver": ["saga"],
        "logistic__l1_ratio": np.linspace(0, 1, 10)  # ElasticNet için l1_ratio ekliyoruz
    },
    {
        "logistic__penalty": ["none"],
        "logistic__solver": ["newton-cg", "lbfgs", "sag", "saga"]
    }
]
```

```python
grid_model = GridSearchCV(model, param_grid= param_grid, cv=5, return_train_score=True)
```
